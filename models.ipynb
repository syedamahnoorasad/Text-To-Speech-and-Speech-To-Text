```python
def speech_to_text(audio, model, processor):  
    trans = []
    for x in range(len(audio)):
        input_values = processor(audio[x], sampling_rate=16000, return_tensors="pt").input_values
        with torch.no_grad():
            logits = model(input_values).logits
        predicted_ids = torch.argmax(logits, dim=-1)
        transcription = processor.batch_decode(predicted_ids)[0]
        print(transcription)
        trans.append(transcription)
    return trans
transcribed_eq = speech_to_text(audio, model2, processor2)
```

```python
from fuzzywuzzy import process

def form_equation(my_string, dictionary):
    keywords = [" plus ", "pls ", " divide ", "dide ", " minus ", " multiply ", "mltiply "]
    num = []
    operand = []
    ans = 0
    
    # Find the keyword used in the string
    my_words = my_string

    for keyword in keywords:
        if keyword in my_string:
            operand = keyword
            my_words = my_string.split(keyword)
            break

    
    # Find the numbers used in the string
    for word in my_words:
        closest_match = process.extractOne(word, dictionary.keys())
        if closest_match[1] >= 70:  # Set a threshold for the matching score
            num.append(dictionary[closest_match[0]])
        else:
            return  # Return None if a number cannot be identified
    
    # Calculate the answer
    if operand == keywords[0] or operand == keywords[1]:
        ans = num[0] + num[1] 
    elif operand == keywords[2] or operand == keywords[3]:
        ans = num[0] / num[1] 
    elif operand == keywords[4]:
        ans = num[0] - num[1] 
    elif operand == keywords[5] or operand == keywords[6]:
        ans = num[0] * num[1]  
    
    # Find the word for the answer
    closest_match = process.extractOne(str(ans), dictionary.values())
    if closest_match[1] >= 80:  # Set a threshold for the matching score
        answer = closest_match[0]
    else:
        return  # Return None if an answer word cannot be identified
    
    eq_in_words = "The answer is " + str(answer) 
    return eq_in_words

# equations_words = []
# for transcripts in transcribed_eq:
#   equations_words.append(form_equation(transcripts, dictionary))

# print(equations_words)
```
